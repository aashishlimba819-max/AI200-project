```python
import re
```

```python
def convert_torque(value):

    if pd.isna(value):
        return np.nan

    value = str(value).lower().replace(',', '')

    number = re.search(r'\d+\.?\d*', value)

    if number is None:
        return np.nan

    torque = float(number.group())

    # Explicit Nm notation
    if 'nm' in value:
        return torque

    # Explicit kgm notation with plausible torque values
    if 'kgm' in value and torque <= 50:
        return torque * 9.80665

    # Ambiguous formats: don't guess
    return np.nan
```

```python
def recover_torque(value):

    if pd.isna(value):
        return np.nan

    value = str(value).lower().replace(',', '')

    # Explicitly identified unusual formats
    if re.match(r'^250@', value):
        return 250.0

    if re.match(r'^510@', value):
        return 510.0

    if re.match(r'^110\(11\.2\)@', value):
        return 110.0

    if re.match(r'^210\s*/\s*1900$', value):
        return 210.0

    return np.nan
```

```python
# Count records and known seat values for each vehicle name

seat_by_name = df_clean.groupby('name').agg(
    total_records=('seats', 'size'),
    known_seats=('seats', 'count'),
    median_seats=('seats', 'median')
)

# Find vehicle names having missing seats
missing_name_info = seat_by_name[
    seat_by_name['known_seats'] > 0
]

print(missing_name_info.head(20))
```

```python
# Vehicle names with at least one known seat value
known_names = df_clean.groupby('name')['seats'].transform(
    lambda x: x.notna().any()
)

recoverable = (
    df_clean['seats'].isna() & known_names
).sum()

print("Missing seats:", df_clean['seats'].isna().sum())
print("Recoverable using same vehicle name:", recoverable)
```

```python
# Calculate median seats for each vehicle name
name_median = df_clean.groupby('name')['seats'].transform('median')

# Identify missing values that can be recovered
mask = df_clean['seats'].isna() & name_median.notna()

# Fill using same vehicle name
df_clean.loc[mask, 'seats'] = name_median[mask]

print("Recovered seat values:", mask.sum())
print("Remaining missing seats:", df_clean['seats'].isna().sum())
```

```python
# Inspect remaining missing seat records

remaining_seats = df_clean[df_clean['seats'].isna()]

print("Remaining records:", len(remaining_seats))

remaining_seats[
    ['name', 'year', 'fuel', 'engine', 'max_power']
].head(30)
```

```python
# Extract manufacturer and model prefix
df_clean['model_group'] = (
    df_clean['name']
    .str.split()
    .str[:2]
    .str.join(' ')
)

# Calculate median seats for each model group
model_median = (
    df_clean.groupby('model_group')['seats']
    .transform('median')
)

# Identify recoverable missing values
model_mask = (
    df_clean['seats'].isna() &
    model_median.notna()
)

print("Recoverable using model group:", model_mask.sum())
```

```python
# Check the number of distinct seat configurations
group_check = df_clean.groupby('model_group').agg(
    known_seats=('seats', 'count'),
    different_seat_values=('seats', 'nunique'),
    median_seats=('seats', 'median')
)

# Groups with multiple seating configurations
ambiguous_groups = group_check[
    group_check['different_seat_values'] > 1
]

print("Ambiguous model groups:", len(ambiguous_groups))
print(ambiguous_groups.sort_values(
    'different_seat_values',
    ascending=False
).head(20))
```

```python
# Number of distinct seat configurations per model group
group_nunique = (
    df_clean.groupby('model_group')['seats']
    .transform('nunique')
)

# Median seats for each model group
group_median = (
    df_clean.groupby('model_group')['seats']
    .transform('median')
)

# Fill only when the group has exactly one known seat configuration
safe_mask = (
    df_clean['seats'].isna() &
    (group_nunique == 1)
)

print("Safely recoverable records:", safe_mask.sum())
print("Remaining missing records:",
      df_clean['seats'].isna().sum() - safe_mask.sum())
```

```python
# Fill missing seats only where model group has one
# consistent known seat configuration

df_clean.loc[safe_mask, 'seats'] = group_median[safe_mask]

print("Recovered values:", safe_mask.sum())
print("Remaining missing seats:", df_clean['seats'].isna().sum())
```
